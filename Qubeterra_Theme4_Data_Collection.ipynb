```python
import requests
import os
import time

def download_pdf(url, paper_id):
    try:
        response = requests.get(
            url,
            timeout=30,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        if response.status_code == 200:
            path = f"papers_pdf/{paper_id}.pdf"

            with open(path, "wb") as f:
                f.write(response.content)

            return path

        return None

    except Exception as e:
        print(f"Error with {paper_id}: {e}")
        return None
```

```python
downloaded = []

for _, row in df.iterrows():

    paper_id = row["paper_id"]
    url = row["paper_url"]

    path = download_pdf(url, paper_id)

    if path:
        downloaded.append(paper_id)

    time.sleep(1)

print("Downloaded:", len(downloaded))
```

```python
texts = []

for pdf_file in pdf_files:

    pdf_path = os.path.join("papers_pdf", pdf_file)

    try:
        doc = fitz.open(pdf_path)

        text = ""

        for page in doc:
            text += page.get_text()

        doc.close()

        paper_id = os.path.splitext(pdf_file)[0]

        texts.append({
            "paper_id": paper_id,
            "text": text
        })

    except Exception as e:
        print("Error:", pdf_file, e)

print("Papers with extracted text:", len(texts))
```

```python
import re

def clean_text(text):
    text = text.replace("\n", " ")
    text = re.sub(r"\s+", " ", text)
    text = re.sub(r"[^\x00-\x7F]+", " ", text)
    return text.strip()
```

```python
cleaned_texts = []

for item in texts:

    cleaned = clean_text(item["text"])

    cleaned_texts.append({
        "paper_id": item["paper_id"],
        "text": cleaned
    })

print("Cleaned papers:", len(cleaned_texts))
```

```python
paper_metadata = df.set_index("paper_id").to_dict("index")

documents = []

for item in cleaned_texts:

    paper_id = item["paper_id"]

    if paper_id in paper_metadata:
        metadata = paper_metadata[paper_id]

        documents.append({
            "paper_id": paper_id,
            "title": metadata["title"],
            "authors": metadata["authors"],
            "year": metadata["year"],
            "debate_topic": metadata["debate_topic"],
            "method": metadata["method"],
            "source": metadata["discovery_sources"],
            "text": item["text"]
        })

print("Documents:", len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)
```

```python
chunks = []

for doc in documents:

    text_chunks = splitter.split_text(doc["text"])

    for i, chunk in enumerate(text_chunks):

        chunks.append({
            "chunk_id": f"{doc['paper_id']}_chunk_{i}",
            "paper_id": doc["paper_id"],
            "title": doc["title"],
            "authors": doc["authors"],
            "year": doc["year"],
            "debate_topic": doc["debate_topic"],
            "method": doc["method"],
            "source": doc["source"],
            "text": chunk
        })

print("Total chunks:", len(chunks))
```

```python
for i in range(3):
    print("\n--- CHUNK", i, "---")
    print(chunks[i]["text"][:500])
```

```python
import json
import os

os.makedirs("papers_chunks", exist_ok=True)

with open("papers_chunks/chunks.json", "w", encoding="utf-8") as f:
    json.dump(chunks, f, ensure_ascii=False, indent=2)

print("Chunks saved successfully")
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
chunk_texts = [chunk["text"] for chunk in chunks]

print("Number of texts:", len(chunk_texts))
```

```python
embeddings = model.encode(
    chunk_texts,
    show_progress_bar=True,
    batch_size=32
)

print("Embedding shape:", embeddings.shape)
```

```python
import numpy as np

np.save(
    "papers_chunks/embeddings.npy",
    embeddings
)

print("Embeddings saved successfully")
```

```python
with open("papers_chunks/chunks_metadata.json", "w", encoding="utf-8") as f:
    json.dump(chunks, f, ensure_ascii=False, indent=2)

print("Metadata saved successfully")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

query = "How do Transformers handle long sequence modeling?"

query_embedding = model.encode([query])

similarities = cosine_similarity(
    query_embedding,
    embeddings
)[0]

top_k = 5
top_indices = np.argsort(similarities)[::-1][:top_k]

print("Top results:\n")

for rank, idx in enumerate(top_indices, 1):
    print(f"\n--- Result {rank} ---")
    print("Similarity:", similarities[idx])
    print("Paper:", chunks[idx]["title"])
    print("Chunk ID:", chunks[idx]["chunk_id"])
    print("Text:", chunks[idx]["text"][:700])
```
